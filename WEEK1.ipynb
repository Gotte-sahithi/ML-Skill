```python
import pandas as pd
# File name
CSV_PATH = "Titanic-Dataset.csv"
# Read CSV file
df = pd.read_csv(CSV_PATH)
```

```python
# 1. Display first 5 rows
print("FIRST 5 ROWS")
print(df.head())
```

```python
# 2. Display dataset shape
print("\nDATASET SHAPE")
print(df.shape)
```

```python
# 3. Display column names
print("\nCOLUMN NAMES")
print(df.columns)
```

```python
# 4. Display data types
print("\nDATA TYPES")
print(df.dtypes)
```

```python
# 5. Display basic information
print("\nDATASET INFORMATION")
df.info()
```

```python
# 6. Display statistical summary
print("\nSTATISTICAL SUMMARY")
print(df.describe())
```

```python
# 7. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())
```

```python
# 8. Check duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())
```

```python
# 9. Survived status distribution
print("\nSURVIVED STATUS DISTRIBUTION")
print(df["Survived"].value_counts())
```

```python
# 10. Survived status percentage
print("\nSURVIVED STATUS PERCENTAGE")
print(df["Survived"].value_counts(normalize=True) * 100)
```
